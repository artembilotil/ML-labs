```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             classification_report, confusion_matrix, ConfusionMatrixDisplay)

VARIANT = 'bean'   
#шлях до даних
DATA = Path('.')
pd.set_option('display.width', 160); pd.set_option('display.max_columns', 20)

if VARIANT == 'bean':
    df = pd.read_csv(DATA / 'drybean.csv')
    target = 'Class'
    features = [c for c in df.columns if c not in ('Class', 'split')]
else:
    df = pd.read_csv(DATA / 'occupancy.csv')
    target = 'Occupancy'
    features = ['Temperature', 'Humidity', 'Light', 'CO2', 'HumidityRatio']

#дані вже розмічені
train, val, test = (df[df['split'] == s] for s in ('train', 'val', 'test'))
X_tr, y_tr = train[features], train[target]
X_va, y_va = val[features], val[target]
X_te, y_te = test[features], test[target]
print('train / validation / test:', len(train), len(val), len(test))
print("\n")

#----------------------------Завдання 1. Попередній аналіз даних---------------------------

print("----------------------------Завдання 1. Попередній аналіз даних---------------------------")
print("\n")

# (а) Перегляд даних
display(X_tr.head())
display(X_tr.describe())

# (б) Розподіл класів
class_counts = y_tr.value_counts()
print("Кількість прикладів кожного класу:\n", class_counts)

plt.figure(figsize=(10, 5))
class_counts.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title("Розподіл класів квасолі у навчальній вибірці (train)")
plt.ylabel("Кількість")
plt.show()

# (в) Точність "лінивої" моделі
most_frequent_class = class_counts.idxmax()
lazy_preds = [most_frequent_class] * len(y_va)
lazy_accuracy = accuracy_score(y_va, lazy_preds)
print(f"Найчастіший клас: {most_frequent_class}")
print(f"Точність лінивої моделі на validation: {lazy_accuracy:.4f}")
print("\n")

#---------------------------Завдання 2. Дерево рішень---------------------------

print("----------------------------Завдання 2. Дерево рішень---------------------------")
print("\n")

# (а) Навчання дерева
dt = DecisionTreeClassifier(max_depth=3, random_state=0)
dt.fit(X_tr, y_tr)

# (б) Точність
print(f"Точність дерева на train: {accuracy_score(y_tr, dt.predict(X_tr)):.4f}")
print(f"Точність дерева на validation: {accuracy_score(y_va, dt.predict(X_va)):.4f}")

# (в) Візуалізація
plt.figure(figsize=(20, 10))
plot_tree(dt, feature_names=features, class_names=dt.classes_, filled=True, rounded=True)
plt.show()

print("Правила дерева (текстом):")
print(export_text(dt, feature_names=features))
print("\n")

#---------------------------Завдання 3. Випадковий ліс---------------------------

print("----------------------------Завдання 3. Випадковий ліс---------------------------")
print("\n")

# (а) Навчання лісу
rf = RandomForestClassifier(n_estimators=100, random_state=0)
rf.fit(X_tr, y_tr)

# (б) Метрики
for name, model in [('Дерево (max_depth=3)', dt), ('Випадковий ліс', rf)]:
    preds = model.predict(X_va)
    acc = accuracy_score(y_va, preds)
    prec = precision_score(y_va, preds, average='macro', zero_division=0)
    rec = recall_score(y_va, preds, average='macro')
    print(f"{name:25s} | Accuracy: {acc:.4f} | Precision: {prec:.4f} | Recall: {rec:.4f}")

# (в) Матриці помилок
fig, axes = plt.subplots(1, 2, figsize=(15, 6))
ConfusionMatrixDisplay.from_estimator(dt, X_va, y_va, ax=axes[0], xticks_rotation=45, cmap='Blues')
axes[0].set_title('Матриця помилок: Дерево')
ConfusionMatrixDisplay.from_estimator(rf, X_va, y_va, ax=axes[1], xticks_rotation=45, cmap='Greens')
axes[1].set_title('Матриця помилок: Випадковий ліс')
plt.show()
print("\n")

#---------------------------Завдання 4. Важливість ознак---------------------------
print("----------------------------Завдання 4. Важливість ознак---------------------------")
print("\n")

# (а) Експеримент з глибиною дерева
depths = range(1, 21)
dt_train_acc, dt_val_acc = [], []

for d in depths:
    m = DecisionTreeClassifier(max_depth=d, random_state=0).fit(X_tr, y_tr)
    dt_train_acc.append(accuracy_score(y_tr, m.predict(X_tr)))
    dt_val_acc.append(accuracy_score(y_va, m.predict(X_va)))

plt.figure(figsize=(12, 5))
plt.plot(depths, dt_train_acc, label='Train Accuracy', marker='o')
plt.plot(depths, dt_val_acc, label='Validation Accuracy', marker='s')
plt.xlabel('Глибина дерева (max_depth)')
plt.ylabel('Точність (Accuracy)')
plt.title('Дерево рішень: Точність від глибини')
plt.xticks(depths)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# (б) Експеримент з кількістю дерев у лісі
n_trees = [1, 5, 10, 25, 50, 100, 200, 300]
rf_val_acc = []

for n in n_trees:
    m = RandomForestClassifier(n_estimators=n, random_state=0, n_jobs=-1).fit(X_tr, y_tr)
    rf_val_acc.append(accuracy_score(y_va, m.predict(X_va)))

plt.figure(figsize=(12, 5))
plt.plot(n_trees, rf_val_acc, marker='o', color='green')
plt.xlabel('Кількість дерев (n_estimators)')
plt.ylabel('Точність (Accuracy) на validation')
plt.title('Випадковий ліс: Точність від кількості дерев')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# (в) Оцінка найкращих моделей на test
best_depth = depths[np.argmax(dt_val_acc)]
print(f"\nНайкраща глибина дерева за validation: {best_depth}")

best_dt = DecisionTreeClassifier(max_depth=best_depth, random_state=0).fit(X_tr, y_tr)
print(f"Точність найкращого дерева на test: {accuracy_score(y_te, best_dt.predict(X_te)):.4f}")
print(f"Точність лісу (n=100) на test: {accuracy_score(y_te, rf.predict(X_te)):.4f}")
print("\n")

#---------------------------Завдання 5. Важливість ознак---------------------------
print("----------------------------Завдання 5. Важливість ознак---------------------------")
print("\n")

# (а) Графіки важливості
fig, axes = plt.subplots(1, 2, figsize=(15, 6))
pd.Series(dt.feature_importances_, index=features).sort_values().plot(kind='barh', ax=axes[0], color='skyblue')
axes[0].set_title('Важливість ознак: Дерево (max_depth=3)')

pd.Series(rf.feature_importances_, index=features).sort_values().plot(kind='barh', ax=axes[1], color='lightgreen')
axes[1].set_title('Важливість ознак: Випадковий ліс')
plt.tight_layout()
plt.show()

# (б) Топ-3 ознаки лісу
top3_features = pd.Series(rf.feature_importances_, index=features).nlargest(3).index.tolist()
print(f"Три найважливіші ознаки лісу: {top3_features}")

# (в) Навчання усіченого лісу
X_tr_top3, X_va_top3, X_te_top3 = X_tr[top3_features], X_va[top3_features], X_te[top3_features]
rf_top3 = RandomForestClassifier(n_estimators=100, random_state=0).fit(X_tr_top3, y_tr)

print(f"Повний ліс (test acc): {accuracy_score(y_te, rf.predict(X_te)):.4f}")
print(f"Ліс на 3 ознаках (test acc): {accuracy_score(y_te, rf_top3.predict(X_te_top3)):.4f}")
print("\n")

#---------------------------Завдання 6. Нові випадки---------------------------
print("----------------------------Завдання 6. Нові випадки---------------------------")
print("\n")

# (а) Створення 3 прикладів
sira_median = X_tr[y_tr == 'SIRA'].median()
dermason_median = X_tr[y_tr == 'DERMASON'].median()
bombay_median = X_tr[y_tr == 'BOMBAY'].median()
midpoint = (sira_median + dermason_median) / 2 # Спірна точка

new_samples = pd.DataFrame([sira_median, midpoint, bombay_median], 
                           index=['SIRA_Median', 'SIRA_DERM_Midpoint', 'BOMBAY_Median'])

# (б) та (в) Прогнози та ймовірності
dt_preds = best_dt.predict(new_samples)
rf_preds = rf.predict(new_samples)
rf_probs = rf.predict_proba(new_samples)

results = pd.DataFrame({
    'Tree_Predict': dt_preds,
    'Forest_Predict': rf_preds,
    'Forest_Probabilities': [str(np.round(p, 3)) for p in rf_probs]
}, index=new_samples.index)

print("Ймовірності виведені у порядку класів:", rf.classes_)
display(results)
print("\n")

#---------------------------Завдання 7. Порахувати руками---------------------------
print("----------------------------Завдання 7. Порахувати руками---------------------------")
print("\n")

# (а) Джині кореня і зменшення ΔG
root_val = dt.tree_.value[0][0]
left_val = dt.tree_.value[1][0]
right_val = dt.tree_.value[dt.tree_.children_right[0]][0]

def calc_gini(val_array):
    p = val_array / val_array.sum()
    return 1 - np.sum(p**2)

gini_root = calc_gini(root_val)
gini_left = calc_gini(left_val)
gini_right = calc_gini(right_val)

n_root = root_val.sum()
gini_split = (left_val.sum()/n_root)*gini_left + (right_val.sum()/n_root)*gini_right
delta_g = gini_root - gini_split

print(f"(а) Gini root: {gini_root:.4f}, Gini left: {gini_left:.4f}, Gini right: {gini_right:.4f}")
print(f"    Зменшення нечистоти (ΔG): {delta_g:.4f}")

# (б) Влучність і повнота SIRA
cm = confusion_matrix(y_va, rf.predict(X_va), labels=rf.classes_)
sira_idx = list(rf.classes_).index('SIRA')
sira_tp = cm[sira_idx, sira_idx]
sira_pred_total = cm[:, sira_idx].sum()
sira_true_total = cm[sira_idx, :].sum()

print(f"\n(б) SIRA - Правильні прогнози (TP): {sira_tp}")
print(f"    SIRA - Всього прогнозів 'SIRA' (стовпець): {sira_pred_total}")
print(f"    SIRA - Всього справжніх 'SIRA' (рядок): {sira_true_total}")
print(f"    Влучність (Precision): {sira_tp / sira_pred_total:.4f}")
print(f"    Повнота (Recall): {sira_tp / sira_true_total:.4f}")

# (в) Бутстреп ймовірність
n = len(X_tr)
p_out_of_bag = 1 - (1 - 1/n)**n
print(f"\n(в) Ймовірність об'єкта не потрапити у вибірку дерева (n={n}): {p_out_of_bag:.4f}")
```
